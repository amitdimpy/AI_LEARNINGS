```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, LabelEncoder
```

```python
plt.scatter(df['mileage'], df['price'])
```

```python
X = df [[ 'mileage', 'age']]
y = df [[ 'price' ]]
```

```python
X
```

```python
y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0) #Train and test on data and test size is 20% and train size is 80%
```

```python
len(X_train) #Length of training data
```

```python
len(X_test)   #Length of testing data
```

```python
X_train  #Chooses data randomly, it keeps changing when you execute it. Use random_state to keep test and train data same
```

```python
X_test
```

```python
reg = LinearRegression()
reg.fit(X_train, y_train)
```

```python
reg.predict(X_test)
```

```python
y_test
```

```python
reg.score(X_test, y_test)
```
