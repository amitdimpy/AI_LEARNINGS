K-Means CLustering Setup

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
from sklearn.cluster import KMeans
from matplotlib import pyplot as plt
%matplotlib inline
import seaborn as sns
from sklearn.datasets import load_iris
iris = load_iris()
```

```python
dir(iris)
```

```python
print(iris.data, iris.feature_names, iris.target ,iris.target_names)
```

```python
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df.head()
```

```python
df1 = df.drop(columns = ['sepal length (cm)','sepal width (cm)'])
df1
```

```python
plt.scatter(df1['petal length (cm)'], df1['petal width (cm)'])
```

```python
km = KMeans()
km
```

```python
y_predicted = km.fit_predict(df1)
y_predicted
```

```python
df1['cluster'] = y_predicted
df1.tail(20)
```

```python
df2 = df1[df1.cluster==0]
df3 = df1[df1.cluster==1]
df4 = df1[df1.cluster==2]
df5 = df1[df1.cluster==3]
df6 = df1[df1.cluster==4]
df7 = df1[df1.cluster==5]
df8 = df1[df1.cluster==6]
df9 = df1[df1.cluster==7]

plt.scatter(df2['petal length (cm)'], df2['petal width (cm)'],color='red')
plt.scatter(df3['petal length (cm)'], df3['petal width (cm)'],color='blue')
plt.scatter(df4['petal length (cm)'], df4['petal width (cm)'],color='green')
plt.scatter(df5['petal length (cm)'], df5['petal width (cm)'],color='yellow')
plt.scatter(df6['petal length (cm)'], df6['petal width (cm)'],color='orange')
plt.scatter(df7['petal length (cm)'], df7['petal width (cm)'],color='purple')
plt.scatter(df8['petal length (cm)'], df8['petal width (cm)'],color='black')
plt.scatter(df9['petal length (cm)'], df9['petal width (cm)'],color='grey')

plt.xlabel('petal length (cm)')
plt.ylabel('petal width (cm)')
plt.show()
```

```python
scalar = MinMaxScaler()
scalar.fit(df1[['petal length (cm)']])
df1['petal length (cm)'] = scalar.transform(df1[['petal length (cm)']])

scalar.fit(df1[['petal width (cm)']])
df1['petal width (cm)'] = scalar.transform(df1[['petal width (cm)']])
df1
```

```python
km = KMeans()
y_predicted = km.fit_predict(df1)
#df1 = df1.drop('cluster', axis=1)
print(y_predicted, df1.head())
```

```python
km.cluster_centers_
```

```python
df2 = df1[df1.cluster==0]
df3 = df1[df1.cluster==1]
df4 = df1[df1.cluster==2]
df5 = df1[df1.cluster==3]
df6 = df1[df1.cluster==4]
df7 = df1[df1.cluster==5]
df8 = df1[df1.cluster==6]
df9 = df1[df1.cluster==7]

plt.scatter(df2['petal length (cm)'], df2['petal width (cm)'],color='red')
plt.scatter(df3['petal length (cm)'], df3['petal width (cm)'],color='blue')
plt.scatter(df4['petal length (cm)'], df4['petal width (cm)'],color='green')
plt.scatter(df5['petal length (cm)'], df5['petal width (cm)'],color='yellow')
plt.scatter(df6['petal length (cm)'], df6['petal width (cm)'],color='orange')
plt.scatter(df7['petal length (cm)'], df7['petal width (cm)'],color='purple')
plt.scatter(df8['petal length (cm)'], df8['petal width (cm)'],color='black')
plt.scatter(df9['petal length (cm)'], df9['petal width (cm)'],color='grey')
plt.scatter(km.cluster_centers_[:,0], km.cluster_centers_[:,1],color='brown', marker='*')

plt.xlabel('petal length (cm)')
plt.ylabel('petal width (cm)')
plt.show()
```
