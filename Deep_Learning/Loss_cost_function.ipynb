```python
import numpy as np
```

Mean Squared Error (MSE)

```python
def mse(y_true, y_pred):
    y_pred=np.array(y_pred)
    y_true=np.array(y_true)
    mse=np.mean((y_true - y_pred)  ** 2)
    return mse
```

```python
actual_values = [1, 1, 0, 2, 4]
predicted_values = [0.3,0.5,0,0.8, 0.9]
result = mse(actual_values, predicted_values)
print (f"Actual Value: {actual_values}, Predicted Value: {predicted_values}, MSE: {result}")
```

Mean Absolute Error (MAE)

```python
def mae(y_true1, y_pred1):
    y_true1=np.array(y_true1)
    y_pred1=np.array(y_pred1)
    mae=np.mean(np.abs(y_true1-y_pred1))
    return mae
```

```python
actual_value1=[1, 1, 2, 3, 0]
predict_value1=[0.6, 0.2, 0.1, 0.4, 0.9]
result1 = mse(actual_value1, predict_value1)
print (f"Actual Value1 : {actual_value1}, Predicted Value : {predict_value1}, MAE: {result1}")
```
