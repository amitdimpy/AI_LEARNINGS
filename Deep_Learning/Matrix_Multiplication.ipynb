```python
import numpy as np
```

```python
revenue = np.array([[800,1000,1000],[50,90,200],[1500,250,350]])
expenses = np.array([[300,500,600],[20,40,60],[100,50,90]])
```

Matrix Broadcasting

```python
profit = np.array([revenue - expenses])
profit
```

```python
price = np.array([50,100,80])
items_sold = np.array([[10,40,400],[50,85,90],[500,600,100]])
```

```python
price*items_sold
```

Dot Product of Matrices

```python
total_price = np.dot(price,items_sold)
total_price
```
