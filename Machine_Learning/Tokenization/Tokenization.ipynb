Tokenization

```python
from transformers import AutoTokenizer
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
```

Tokenizing simple sentences

```python
text1 = "My name is Amit and I am a software engineer and an AI enthusiast."
tokens1 = tokenizer.tokenize(text1)
print(f"Tokens: {tokens1}")
tokenid1 = tokenizer.encode(text1)
print(f"Token ids: {tokenid1}")
```

Tokeninzing complex sentences

```python
text2 = "I was flabbergasted by watching his scintillating performance."
tokens2 = tokenizer.tokenize(text2)
print(f"Tokens: {tokens2}")
tokenid2 = tokenizer.encode(text2)
print(f"Token ids: {tokenid2}")
```

Tokenizing large sentences

```python
text3 = ("Professional Summary"
         "Experienced professional with 20 years of experience in the IT industry, specializing in Azure Cloud, Data Engineering, and Infrastructure Architecture. Proven expertise in designing secure, scalable cloud solutions, managing enterprise data platforms, and implementing governance frameworks. Strong background in DevOps enablement, automation, and data integration, with hands-on experience across cloud, AI/ML, and legacy data center environments."
         "Core Competencies"
         "Azure Cloud Architecture & Governance"
         "Data Engineering & Data Pipelines"
         "DevOps & Infrastructure Automation"
         "Security & Access Management (IAM)"
         "Databricks & Data Lake Governance"
         "AI/ML Model Development & Deployment"
         "Enterprise Storage & Backup Solutions"
         "Network & VNet Architecture"
         "Scripting (Python, Shell, PowerShell)")
tokens3 = tokenizer.tokenize(text3)
print(f"Tokens:{tokens3}")
tokenid3 = tokenizer.encode(text3)
print(f"Token ids:{tokenid3}")
```
