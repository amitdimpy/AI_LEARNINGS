```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import seaborn as sns
from sklearn import linear_model
```

```python
df = pd.read_csv('C:/Users/amitd/OneDrive/Desktop/AI_LEARNINGS/area_price.csv')
df
```

```python
plt.scatter(df.area,df.price)
```

```python
plt.xlabel('area(sq ft)')
plt.ylabel('price(US $)')
plt.scatter(df.area,df.price,color='red',marker='+')
```

```python
reg = linear_model.LinearRegression()
reg.fit(df[['area']],df.price)
```

```python
reg.predict([[3300]])
```

```python
reg.coef_
```

```python
reg.intercept_
```

```python
reg.coef_*3300+reg.intercept_   #output of y=mx+b
```
