```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
from sklearn.decomposition import PCA
```

```python
np.random.seed(42)
n_samples=1000
n_features=50
X_high_dim=np.random.randn(n_samples,n_features)
```

```python
n_components=2
pca=PCA(n_components=n_components)
X_pca=pca.fit_transform(X_high_dim)
```

```python
fig, axs=plt.subplots(1,2, figsize=(12,6))
axs[0].scatter(X_high_dim[:,0], X_high_dim[:,1], c='blue', alpha=0.5)
axs[0].set_xlabel('Feature 1')
axs[0].set_ylabel('Feature 2')
axs[0].set_title('Original Data')

axs[1].scatter(X_pca[:,0], X_pca[:,1], c='red', alpha=0.5)
axs[1].set_xlabel('PCA 1')
axs[1].set_ylabel('PCA 2')
axs[1].set_title('PCA Data')
plt.tight_layout()
plt.show()
```
