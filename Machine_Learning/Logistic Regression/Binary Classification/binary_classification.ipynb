Logistic Regression for Binary Classification

```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
%matplotlib inline
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_csv(r'C:\Users\amitd\OneDrive\Documents\GitHub\AI_LEARNINGS\Logistic Regression\Binary Classification\insurance.csv')
df
```

```python
plt.scatter(df['age'],df['bought_insurance'],marker='+',color='red')
plt.xlabel('Age')
plt.ylabel('Insurance')
```

```python
df.shape  #Shows rows, columns
```

```python
X_train, X_test, y_train, y_test = train_test_split(df[['age']], df.bought_insurance, test_size = 0.2, random_state = 0)
```

```python
X_train
```

```python
y_train
```

```python
print(X_train.shape, X_test.shape, y_train.shape, y_test.shape)
```

```python
model = LogisticRegression()
model.fit(X_train, y_train)
```

```python
X_test
```

```python
model.predict(X_test)  #Prediction is based on the output of X_test
```

```python
model.score(X_test, y_test)
```

```python
model.predict_log_proba(X_test)  #First column is not buy and second is will buy insurance
```

```python
pred1 = model.predict([[10]])
pred2 = model.predict([[70]])
pred3 = model.predict([[40]])
print(pred1, pred2, pred3)
```
