```python
import pandas as pd
import numpy as np
import matplotlib as plt
%matplotlib inline
from sklearn.model_selection import train_test_split
from sklearn.datasets import fetch_20newsgroups #MultinomialNB
from sklearn.feature_extraction.text import CountVectorizer #MultinomialNB
from sklearn.naive_bayes import GaussianNB
from sklearn.naive_bayes import BernoulliNB
from sklearn.naive_bayes import MultinomialNB
from sklearn.datasets import make_classification
from sklearn.metrics import accuracy_score
```

```python
X, y = make_classification(n_samples=1000, n_features=20, n_classes=2, random_state=7)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=7)
```

Gaussian NB

```python
clf = GaussianNB()
clf.fit(X_train, y_train)
y_pred = clf.predict(X_test)
print("Gaussian Naive Bayes:", y_pred)
accclf = accuracy_score(y_test, y_pred)
print("Gaussian Naive Bayes:", y_pred, "Accuracy:", accclf)
```

Bernoulli NB

```python
bnb = BernoulliNB()
bnb.fit(X_train, y_train)
y_pred_bnb = bnb.predict(X_test)
accbnb = accuracy_score(y_test, y_pred_bnb)
print("Bernoulli Naive Bayes:", y_pred_bnb, "Accuracy:", accbnb)
```

Multinomial NB

```python
data = fetch_20newsgroups(subset='all')
vectorizer = CountVectorizer()
X1 = vectorizer.fit_transform(data.data)
y1 = data.target
X_train, X_test, y_train, y_test = train_test_split(X1, y1, test_size=0.2, random_state=42)
```

```python
mnb = MultinomialNB()
mnb.fit(X_train, y_train)
y_pred_mnb = mnb.predict(X_test)
accmnb = accuracy_score(y_test, y_pred_mnb)
print("Multinomial Naive Bayes:", y_pred_mnb, "Accuracy:", accmnb)
```
