Embeddings

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
```

```python
model = SentenceTransformer('sentence-transformers/all-mpnet-base-v2')

# Encode some texts
sentences = [
    "The weather is lovely today.",
    "It's so sunny outside!",
    "He drove to the stadium.",
]
embeddings = model.encode(sentences)
print(embeddings.shape)
similarities = cosine_similarity(embeddings)
print(f"Similarities shape: {similarities.shape}")
print(f"Similarities: {similarities}")
```

```python
model1 = SentenceTransformer('all-MiniLM-L6-v2')
sentences1 = [
    "My name is Amit.",
    "I am learning AI.",
    "I want to make my career in AI."
    "AI is complex and requires continuous practice.",
    "Practice makes the man perfect."]
embeddings1 = model1.encode(sentences1)
print(embeddings1.shape)
similarities1 = cosine_similarity(embeddings1)
print(f"Similarities shape: {similarities1.shape}")
print(f"Similarities: {similarities1}")
```
